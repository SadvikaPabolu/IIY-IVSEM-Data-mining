```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv(r"C:\Users\sadvika\Desktop\Data mining\Stu Performance.csv")

# Remove extra spaces in column names
df.columns = df.columns.str.strip()

# Print column names (to verify)
print("Columns in dataset:")
print(df.columns)

# Select first two numeric columns automatically
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

x_attr = numeric_cols[0]
y_attr = numeric_cols[1]

print("Selected Attributes:", x_attr, "and", y_attr)

# ---------------------------------
# 1️⃣ Scatter Plot
# ---------------------------------
plt.figure()
plt.scatter(df[x_attr], df[y_attr])
plt.title(f"Scatter Plot of {x_attr} vs {y_attr}")
plt.xlabel(x_attr)
plt.ylabel(y_attr)
plt.show()

# ---------------------------------
# 2️⃣ Histogram
# ---------------------------------
plt.figure()
plt.hist(df[x_attr].dropna(), bins=15, alpha=0.5, label=x_attr)
plt.hist(df[y_attr].dropna(), bins=15, alpha=0.5, label=y_attr)
plt.title("Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.legend()
plt.show()

# ---------------------------------
# 3️⃣ Box Plot (Updated – No Warning)
# ---------------------------------
plt.figure()
plt.boxplot([df[x_attr].dropna(), df[y_attr].dropna()],
            tick_labels=[x_attr, y_attr])
plt.title("Box Plot")
plt.ylabel("Values")
plt.show()
```
